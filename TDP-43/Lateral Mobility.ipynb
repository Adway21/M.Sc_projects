```python
import pandas as pd

def lateral_mobility(u,savename):
    p = u.select_atoms("protein")
    p_com = p.center_of_geometry()
    xi = p_com[0]
    yi = p_com[1]
    m = []
    timeframe = []
    for ts in u.trajectory:
        p = u.select_atoms("protein")
        p_com = p.center_of_geometry()
        xn = p_com[0]
        yn = p_com[1]
        distance = np.sqrt((xn - xi)**2 + (yn - yi)**2)
        m.append(distance)
        timeframe.append(ts.frame)
        xi = xn
        yi = yn

    df = pd.DataFrame()
    df['timeframe'] = timeframe
    df['lateral_mobility'] = m
    df.to_csv(savename + '.csv', sep='\t', index=False, header=True)
    return 'lateral_mobility.csv file contains the lateral mobility of protein'
```
