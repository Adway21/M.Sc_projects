```python
import numpy as np
import pandas as pd
import MDAnalysis as mda
import matplotlib.pyplot as plt
```

```python
def tilt_angle(gro_file, xtc_file, initial, final):
    u = mda.Universe(gro_file, xtc_file)
    t1 = []
    time_frames = []
    for ts in u.trajectory:
        time_frames.append(ts.time)
        f = u.select_atoms('resid ' + str(initial) + ' and name CA')
        f_x, f_y, f_z = f.positions[0]
        
        l = u.select_atoms('resid ' + str(final) + ' and name CA')
        l_x, l_y, l_z = l.positions[0]
        
        a1 = np.arccos((f_z - l_z) / np.sqrt((f_x - l_x)**2 + (f_y - l_y)**2 + (f_z - l_z)**2))
        t1.append(np.degrees(a1))
    
    df = pd.DataFrame({'Time (ps)': time_frames, 'Tilt Angle (degrees)': t1})
    df.to_csv('tilt_angle_vs_time.csv', index=False)
    
    plt.figure(figsize=(10, 6))
    plt.plot(time_frames, t1, color='blue')
    plt.xlabel('Time (ps)')
    plt.ylabel('Tilt Angle (degrees)')
    plt.title('Tilt Angle vs Time')
    plt.grid(True)
    #plt.savefig('tilt_angle_vs_time.png')
    plt.show()
```
