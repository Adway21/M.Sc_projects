```python
def rolling_chi1(df1,df2,savename):
    fig, ax = plt.subplots(figsize=(10, 7), dpi=300)
    plt.plot(df1['Time'], df1['Angle of Phe20'], color="blue", alpha=0.05)
    rolling_avg_df1 = df1['Angle of Phe20'].rolling(window=200).mean()
    rolling_avg_df2 = df2['Angle of Phe20'].rolling(window=200).mean()
    plt.plot(df1['Time'], rolling_avg_df1, color="blue", label=r"without Ca$^{2+}$")
    plt.plot(df2['Time'], df2['Angle of Phe20'], color="red", alpha=0.05)
    plt.plot(df2['Time'], rolling_avg_df2, color="red", label=r"with Ca$^{2+}$")
    ax.set_xlabel("Time (ns)", fontsize=22)
    plt.xticks(fontsize=20)
    ax.set_ylabel("Angle of $\chi_{1}$ of Phe20", fontsize=22)
    plt.yticks(fontsize=20)
    plt.tight_layout()
    #plt.legend(loc="lower right",fontsize=20)
    plt.savefig(savename, dpi=500)
    plt.show()
```

```python
rolling_chi1(neutral_chi1,neutral_ca_chi1,"neutral_chi1_Phe20.pdf")
```

```python
from scipy.signal import find_peaks
import scipy.stats as stats
import numpy as np
def detect_and_plot_peaks(df):
    g = df['Angle of Phe26']

    sns.kdeplot(g, label="Data", bw_adjust=1.5)

    kde = stats.gaussian_kde(g)

    x_values = np.linspace(min(g), max(g), 4000)
    peaks, _ = find_peaks(kde(x_values), height=0.001)
    peak_xs = x_values[peaks]

    for peak_x in peak_xs:
        plt.axvline(x=peak_x, color='r', linestyle='--', label=f'Peak: {peak_x:.3f}')

    plt.legend()
    plt.show()

    for peak_x in peak_xs:
        rounded_peak_x = round(peak_x, 3)
        matching_rows = df[df['Angle of Phe26'].round(3) == rounded_peak_x]
        if not matching_rows.empty:
            peak_time = matching_rows['Time'].values[0]
            print(f"Corresponding Time value for peak Cosine value {rounded_peak_x:.3f}: {peak_time}")
        else:
            print(f"No matching rows found for peak Cosine value {rounded_peak_x:.3f}")
'''
    for peak_x in peak_xs:
        peak_time = df.loc[df['Angle of Phe26'].round(3) == peak_x.round(3), 'Time'].values[0]
        print(f"Corresponding Time value for peak Cosine value {peak_x:.3f}: {peak_time}")
'''  
```

```python
import seaborn as sns
import numpy as np
from scipy.signal import find_peaks
import matplotlib.pyplot as plt

g = five_ca_chi1

sns.kdeplot(g['Angle of Phe20'])
ax = plt.gca()
kde_curve_data = ax.lines[0].get_data()
peaks, _ = find_peaks(kde_curve_data[1])
peak_x_values = kde_curve_data[0][peaks]
print("X-values of peaks:", [f"{x:.3f}" for x in peak_x_values])
plt.show()
corresponding_times = []

for peak_x in peak_x_values:
    nearest_time = g.loc[
        (g['Angle of Phe20'] - peak_x).abs().idxmin(), 'Time'
    ]
    corresponding_times.append(nearest_time)
print("Corresponding time values for each peak x-value:", corresponding_times)
```
