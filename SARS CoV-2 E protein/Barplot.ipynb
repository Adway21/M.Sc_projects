```python
def kl(p,q):
    p += abs(np.min(p)) + 1e-12
    q += abs(np.min(q)) + 1e-12
    p /= np.sum(p)
    q /= np.sum(q)
    return np.sum(p * np.log(p / q))
```

```python
def zzzzz(p,q):
    p_smooth = p + 1e-12
    q_smooth = q + 1e-12
    return np.sum(p_smooth*np.log(p_smooth/q_smooth))
```
